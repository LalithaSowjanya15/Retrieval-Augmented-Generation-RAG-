### **9) Sum of Numbers from 1 to N**
##### **Problem Statement:**
Take a number N and find the sum from 1 to N.  
##### **Example:**
Input: 5

1 + 2 + 3 + 4 + 5 = 15

```python
Number = int(input("Enter a number to calculate sum of numbers from 1 to N: "))
Sum = 0
for i in range(1, Number+1):
  Sum = Sum + i
print(Sum)
```

### **10) Count Even and Odd Numbers**
##### **Problem Statement:**
Take a number N and count:
* Total Even Numbers
* Total Odd Numbers

between 1 and N.
##### **Example:**
Input: 10

Even Count = 5

Odd Count = 5

```python
N = int(input("Enter a number to count Even and Odd Numbers from 1 to N: "))
Even_count = 0
Odd_count =0

for i in range(1, N+1):
  if i % 2 == 0:
    Even_count += 1
  else:
    Odd_count += 1
print("Even count: ", Even_count)
print("Odd count: ", Odd_count)
```
